### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
# Keep relevant columns and drop missing values
comp_df = df[['Age', 'ConvertedCompYearly']].dropna()

# Define an order for age groups (important for line charts)
age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

# Convert Age to ordered categorical
comp_df['Age'] = pd.Categorical(comp_df['Age'], categories=age_order, ordered=True)

# Calculate median compensation by age group
median_comp = (
    comp_df
    .groupby('Age', observed=True)['ConvertedCompYearly']
    .median()
    .reset_index()
)
##Create the Line Chart
plt.figure(figsize=(10, 5))
plt.plot(
    median_comp['Age'],
    median_comp['ConvertedCompYearly'],
    marker='o',
    linewidth=2
)

plt.title('Median Yearly Compensation by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Median ConvertedCompYearly')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
##Median yearly compensation rises with age, peaks in mid-career, and stabilizes or slightly declines in later age groups.
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
## Write your code here
# Filter age groups between 25 and 45
age_focus = df[df['Age'].isin([
    '25-34 years old',
    '35-44 years old'
])][['Age', 'ConvertedCompYearly']].dropna()

# Ensure correct order
age_order = ['25-34 years old', '35-44 years old']
age_focus['Age'] = pd.Categorical(age_focus['Age'], categories=age_order, ordered=True)

# Compute median compensation
median_comp = (
    age_focus
    .groupby('Age', observed=True)['ConvertedCompYearly']
    .median()
    .reset_index()
)

median_comp
## Create the Line Chart

plt.figure(figsize=(8, 5))

plt.plot(
    median_comp['Age'],
    median_comp['ConvertedCompYearly'],
    marker='o',
    linewidth=2
)

plt.title('Median Yearly Compensation (Ages 25–45)')
plt.xlabel('Age Group')
plt.ylabel('Median ConvertedCompYearly')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
##Median yearly compensation increases noticeably between ages 25–34 and 35–44, reflecting early to mid-career salary growth.
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here
# Select relevant columns and drop missing values
exp_df = df[['YearsCodePro', 'JobSatPoints_6']].dropna()

# Convert YearsCodePro to numeric (handles 'Less than 1 year', 'More than 50 years', etc.)
exp_df['YearsCodePro'] = pd.to_numeric(exp_df['YearsCodePro'], errors='coerce')

# Drop rows where conversion failed
exp_df = exp_df.dropna(subset=['YearsCodePro'])
## Create Experience Bands (Recommended) -This smooths the trend and makes the line chart meaningful.

# Create experience bands
exp_df['ExperienceBand'] = pd.cut(
    exp_df['YearsCodePro'],
    bins=[0, 2, 5, 10, 20, 30, 50],
    labels=['0–2', '3–5', '6–10', '11–20', '21–30', '30+']
)
## Aggregate Median Job Satisfaction
 
median_sat = (
    exp_df
    .groupby('ExperienceBand', observed=True)['JobSatPoints_6']
    .median()
    .reset_index()
)

## Line Chart: Job Satisfaction by Experience Level
 
plt.figure(figsize=(9, 5))

plt.plot(
    median_sat['ExperienceBand'],
    median_sat['JobSatPoints_6'],
    marker='o',
    linewidth=2
)

plt.title('Median Job Satisfaction by Experience Level')
plt.xlabel('Years of Professional Experience')
plt.ylabel('Median Job Satisfaction (JobSatPoints_6)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
##Median job satisfaction increases with professional experience up to mid-career levels and then stabilizes at higher experience bands.
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
## Write your code here
# Select relevant columns
exp_df = df[['YearsCodePro', 'ConvertedCompYearly']].dropna()

# Convert YearsCodePro to numeric
exp_df['YearsCodePro'] = pd.to_numeric(exp_df['YearsCodePro'], errors='coerce')

# Drop rows where conversion failed
exp_df = exp_df.dropna(subset=['YearsCodePro'])
## Create Experience Bands (Recommended)
 
exp_df['ExperienceBand'] = pd.cut(
    exp_df['YearsCodePro'],
    bins=[0, 2, 5, 10, 20, 30, 50],
    labels=['0–2', '3–5', '6–10', '11–20', '21–30', '30+']
)
## Compute Median Compensation by Experience
 
median_comp = (
    exp_df
    .groupby('ExperienceBand', observed=True)['ConvertedCompYearly']
    .median()
    .reset_index()
)

median_comp
## Line Chart: Median Compensation vs Experience

plt.figure(figsize=(9, 5))

plt.plot(
    median_comp['ExperienceBand'],
    median_comp['ConvertedCompYearly'],
    marker='o',
    linewidth=2
)

plt.title('Median Yearly Compensation by Experience Level')
plt.xlabel('Years of Professional Experience')
plt.ylabel('Median ConvertedCompYearly')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
##Median compensation increases with experience, rising rapidly in early and mid-career stages before leveling off at higher experience levels.
```

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.
