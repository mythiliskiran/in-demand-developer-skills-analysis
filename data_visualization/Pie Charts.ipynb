### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
# Drop missing values
db_df = df['DatabaseWantToWorkWith'].dropna()

# Split multiple databases and flatten
db_list = db_df.str.split(';').explode()

# Get top 5 most desired databases
top5_dbs = db_list.value_counts().head(5)
plt.figure(figsize=(7, 7))

plt.pie(
    top5_dbs.values,
    labels=top5_dbs.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top 5 Most Desired Databases to Work With')
plt.axis('equal')  # Makes the pie a circle
plt.show()
##The pie chart shows that PostgreSQL is the most desired database to work with, while interest is fairly evenly distributed among several other popular database technologies.
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
dev_df = df['DevType'].dropna()

# Split multiple roles and flatten
dev_list = dev_df.str.split(';').explode()

# Get top 5 developer roles
top5_devtypes = dev_list.value_counts().head(5)
plt.figure(figsize=(7, 7))

plt.pie(
    top5_devtypes.values,
    labels=top5_devtypes.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top 5 Developer Roles')
plt.axis('equal')  # Ensures the pie chart is circular
plt.show()
##The pie chart shows that full-stack and back-end developers dominate the respondent population, indicating strong demand for versatile and server-side development skills.
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
# Drop missing values
os_df = df['OpSysProfessional use'].dropna()

# Split multiple OS values and flatten
os_list = os_df.str.split(';').explode()

# Get top operating systems
top_os = os_list.value_counts().head(5)
plt.figure(figsize=(7, 7))

plt.pie(
    top_os.values,
    labels=top_os.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top Operating Systems Used Professionally')
plt.axis('equal')  # Ensures circular pie
plt.show()
##The pie chart shows that Windows, Linux, and macOS dominate professional development environments, reflecting diverse platform usage across the industry.
```

### Task 2: Additional Visualizations and Comparisons


##### 2.1 Pie Chart for Top 5 Programming Languages Respondents Have Worked With


The `LanguageHaveWorkedWith` column contains the programming languages that respondents have experience with. We’ll plot a pie chart to display the composition of the top 5 languages.



```python
##Write your code here
# Drop missing values
lang_df = df['LanguageHaveWorkedWith'].dropna()

# Split multiple languages and flatten
lang_list = lang_df.str.split(';').explode()

# Get top 5 most used languages
top5_languages = lang_list.value_counts().head(5)
plt.figure(figsize=(7, 7))

plt.pie(
    top5_languages.values,
    labels=top5_languages.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top 5 Programming Languages Worked With')
plt.axis('equal')  # Ensures a perfect circle
plt.show()
##The pie chart shows that developer experience is concentrated around a small set of core languages, with JavaScript leading, followed closely by HTML/CSS, Python, and SQL.
```

##### 2.2 Pie Chart for Top Collaboration Tools used in Professional Use


Using the `NEWCollabToolsHaveWorkedWith` column, we’ll identify and visualize the top collaboration tools respondents use in their professional work.



```python
##Write your code here
# Drop missing values
tools_df = df['NEWCollabToolsHaveWorkedWith'].dropna()

# Split multiple tools and flatten into one list
tool_list = tools_df.str.split(';').explode()

# Get top 5 collaboration tools
top5_tools = tool_list.value_counts().head(5)
plt.figure(figsize=(7, 7))

plt.pie(
    top5_tools.values,
    labels=top5_tools.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top Collaboration Tools Used Professionally')
plt.axis('equal')  # Keeps pie chart circular
plt.show()
##The pie chart shows that Visual Studio Code is the most widely used professional collaboration tool, with other IDEs and editors sharing the remaining usage.
```

##### 3.1 Pie Chart of `Respondents` Most Admired Programming Languages


The `LanguageAdmired` column lists the programming languages respondents admire most. Create a pie chart to visualize the top 5 admired languages.



```python
##Write your code here
# Drop missing values
admired_df = df['LanguageAdmired'].dropna()

# Split multiple languages into separate rows
admired_list = admired_df.str.split(';').explode()

# Get the top 5 admired languages
top5_admired = admired_list.value_counts().head(5)

plt.figure(figsize=(7, 7))

plt.pie(
    top5_admired.values,
    labels=top5_admired.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Top 5 Admired Programming Languages')
plt.axis('equal')  # Ensures the pie chart is circular
plt.show()
##The pie chart shows that developer admiration is fairly evenly distributed among core programming languages, with JavaScript, Python, and SQL slightly leading.
```
